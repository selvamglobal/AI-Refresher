```python
# Exercise 15 — Grade classifier
score = 78

# conditional operators = >, <, >=, <=, ==, !=

if score >= 90:
    grade = "S"
elif score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Score : {score}")
print(f"Grade : {grade}")

# Practice with different scores
test_scores = [95, 85, 75, 65, 55]
for test_score in test_scores:
    if test_score >= 90:
        result = "S"
    elif test_score >= 80:
        result = "A"
    elif test_score >= 70:
        result = "B"
    elif test_score >= 60:
        result = "C"
    else:
        result = "F"
    print(f"Score: {test_score} → Grade: {result}")
```

```python
# Exercise 16 — API status handler
  
http_codes = [200, 400, 429, 500, 404, 403]
for code in http_codes:
    if code == 200:
        message = "Success — process the response"
    elif code == 400:
        message = "Bad request — check your prompt"
    elif code == 429:
        message = "Rate limit hit — wait and retry"
    elif code == 500:
        message = "Server error — try again later"
    elif code == 404:
        message = "Not found — resource unavailable"
    elif code == 403:
        message = "Forbidden — access denied"
    else:
        message = f"Unknown status: {code}"
    print(f"Status {code}: {message}")
```

```python
# Exercise 2 — Class with multiple methods
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def result(self):
        return "Passed" if self.score >= 60 else "Failed"

    def grade(self):
        if self.score >= 90: return "S"
        elif self.score >= 80: return "A"
        elif self.score >= 70: return "B"
        elif self.score >= 60: return "C"
        else: return "F"

    def display(self):
        print(f"{self.name} | Score: {self.score} | Grade: {self.grade()} | {self.result()}")

s1 = Student("Arun", 88)
s2 = Student("Karan", 45)
s3 = Student("Priya", 95)

s1.display()
s2.display()
s3.display()

# Practice 
students = [
    Student("Meena", 76),
    Student("Ramesh", 60),
    Student("Sathish", 91),
    Student("Anu", 52),
    Student("Dinesh", 84)
]

print("\nAdditional results:")
for student in students:
    student.display()

print("\nQuick summary:")
for student in students:
    print(f"{student.name}: {student.result()} | {student.grade()}")
```

```python
# Exercise 3 — Practice: StudentAccount class (similar concept to BankAccount)
class StudentAccount:
    def __init__(self, student_name, credits=0):
        self.student_name = student_name
        self.credits = credits
        self.activities = []

    def earn_credits(self, amount, source="assignment"):
        self.credits += amount
        self.activities.append(f"Earned   : +{amount} ({source})")
        print(f"Earned {amount} credits from {source}. Total: {self.credits}")

    def use_credits(self, amount, purpose="exam_retake"):
        if amount > self.credits:
            print("Insufficient credits")
        else:
            self.credits -= amount
            self.activities.append(f"Used     : -{amount} ({purpose})")
            print(f"Used {amount} credits for {purpose}. Remaining: {self.credits}")

    def show_history(self):
        print(f"\n--- Credit History for {self.student_name} ---")
        for activity in self.activities:
            print(f"  {activity}")
        print(f"  Current Credits: {self.credits}")

acc1 = StudentAccount("Ramesh", 50)
acc1.earn_credits(20, "quiz")
acc1.use_credits(15, "lab_fee")
acc1.earn_credits(10, "project")
acc1.use_credits(100, "workshop")
acc1.show_history()
```

```python
# Exercise 4 — Library book management
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

    def borrow(self, student_name):
        if self.is_available:
            self.is_available = False
            print(f"'{self.title}' borrowed by {student_name}")
        else:
            print(f"'{self.title}' is not available")

    def return_book(self):
        self.is_available = True
        print(f"'{self.title}' returned. Available again.")

    def status(self):
        state = "Available" if self.is_available else "Borrowed"
        print(f"'{self.title}' by {self.author} — {state}")

b1 = Book("Deep Learning", "Goodfellow")
b2 = Book("Python Crash Course", "Matthes")

b1.status()
b1.borrow("Priya")
b1.borrow("Karan")
b1.return_book()
b1.borrow("Karan")
b2.status()

# Practice 
print("\nAdditional library practice:")
b3 = Book("Introduction to Algorithms", "Cormen")
b4 = Book("Clean Code", "Martin")
b5 = Book("Eloquent JavaScript", "Haverbeke")

b3.status()
b3.borrow("Ramesh")
b3.borrow("Suresh")
b4.status()
b4.borrow("Sathish")
b4.return_book()
b4.borrow("Dinesh")
b5.borrow("Anu")
b5.status()

print("\nFinal statuses:")
b3.status()
b4.status()
b5.status()
```

```python
# Exercise 5 — Simple AI Chatbot session class
class ChatSession:
    def __init__(self, model_name):
        self.model_name = model_name
        self.messages = []
        self.token_count = 0

    def send(self, prompt):
        tokens = len(prompt.split())
        self.token_count += tokens
        self.messages.append({"role": "user", "content": prompt})
        print(f"[{self.model_name}] Prompt received: '{prompt}'")

    def get_history(self):
        print(f"\n--- Chat History ({self.model_name}) ---")
        for i, msg in enumerate(self.messages, start=1):
            print(f"  {i}. {msg['role'].upper()}: {msg['content']}")

    def stats(self):
        print(f"\nTotal messages : {len(self.messages)}")
        print(f"Total tokens   : {self.token_count}")

session = ChatSession("llama3.2")
session.send("What is AI?")
session.send("Explain machine learning")
session.send("What is backpropagation?")
session.get_history()
session.stats()

# Practice 
print("\n" + "="*50)
session2 = ChatSession("mistral")
session2.send("How does Python work?")
session2.send("Explain object-oriented programming")
session2.send("What are classes and objects?")
session2.get_history()
session2.stats()
```

```python
# Exercise 6 — Todo task manager
class TaskManager:
    def __init__(self, owner):
        self.owner = owner
        self.tasks = []

    def add(self, task):
        self.tasks.append({"task": task, "done": False})
        print(f"Added: '{task}'")

    def complete(self, index):
        self.tasks[index]["done"] = True
        print(f"Completed: '{self.tasks[index]['task']}'")

    def pending(self):
        return [t for t in self.tasks if not t["done"]]

    def display(self):
        print(f"\n--- Tasks for {self.owner} ---")
        for i, t in enumerate(self.tasks):
            status = "✓" if t["done"] else "○"
            print(f"  [{status}] {i}. {t['task']}")

tm = TaskManager("Arun")
tm.add("Complete Python exercises")
tm.add("Watch ML lecture")
tm.add("Set up VS Code")
tm.complete(0)
tm.complete(2)
tm.display()
print(f"\nPending: {[t['task'] for t in tm.pending()]}")

# Practice  
print("\n" + "="*50)
tm_priya = TaskManager("Priya")
tm_priya.add("Review numpy basics")
tm_priya.add("Complete assignment 3")
tm_priya.add("Prepare for quiz")
tm_priya.add("Study file I/O operations")
tm_priya.complete(1)
tm_priya.complete(3)
tm_priya.display()
print(f"Remaining tasks: {len(tm_priya.pending())}")

# Practice  
print("\n" + "="*50)
tm_ramesh = TaskManager("Ramesh")
tm_ramesh.add("Train model on dataset")
tm_ramesh.add("Document API endpoints")
tm_ramesh.add("Debug connection error")
tm_ramesh.add("Optimize database queries")
tm_ramesh.add("Write test cases")
tm_ramesh.complete(0)
tm_ramesh.complete(2)
tm_ramesh.complete(4)
tm_ramesh.display()
print(f"Pending tasks: {[t['task'] for t in tm_ramesh.pending()]}")
```

```python
# Exercise 7 — Base class and child classes
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def speak(self):
        print(f"{self.name} says {self.sound}")

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name, "Woof")

    def fetch(self):
        print(f"{self.name} fetches the ball!")

class Cat(Animal):
    def __init__(self, name):
        super().__init__(name, "Meow")

    def purr(self):
        print(f"{self.name} purrs...")

d = Dog("Bruno")
c = Cat("Whiskers")

d.speak()
d.fetch()
c.speak()
c.purr()

# Practice
print("\n" + "="*50)

class Model:
    def __init__(self, name, tokens):
        self.name = name
        self.tokens = tokens

    def info(self):
        print(f"Model: {self.name} | Max tokens: {self.tokens}")

class LargeModel(Model):
    def __init__(self, name):
        super().__init__(name, 8192)

    def optimize(self):
        print(f"{self.name} optimized for complex tasks")

class SmallModel(Model):
    def __init__(self, name):
        super().__init__(name, 2048)

    def quick_response(self):
        print(f"{self.name} responds quickly with limited context")

large = LargeModel("Llama 3.2")
small = SmallModel("Qwen 2.5")

large.info()
large.optimize()
small.info()
small.quick_response()
```

```python
# Exercise 8 — Student → AIStudent inheritance
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def result(self):
        return f"{self.name}: {'Passed' if self.score >= 60 else 'Failed'}"

class AIStudent(Student):
    def __init__(self, name, score, project, model_used):
        super().__init__(name, score)
        self.project = project
        self.model_used = model_used

    def showcase(self):
        print(f"{self.name} built '{self.project}' using {self.model_used}")

    def full_report(self):
        print(f"Name      : {self.name}")
        print(f"Score     : {self.score}")
        print(f"Result    : {self.result()}")
        print(f"Project   : {self.project}")
        print(f"Model     : {self.model_used}")

s = AIStudent("Priya", 95, "RAG Chatbot", "llama3.2")
s1 = AIStudent("Ravi", 90, "RAG Chatbot", "Gemma")
s.showcase()
s.full_report()

s1.showcase()
s1.full_report()

# Practice
print("\n" + "="*50)
ai_students = [
    AIStudent("Karan", 88, "Image Classification", "mistral"),
    AIStudent("Meena", 92, "Sentiment Analysis", "qwen2.5:1.5b"),
    AIStudent("Sathish", 85, "Text Summarizer", "llama3.2"),
    AIStudent("Anu", 78, "Question Answering System", "Gemma")
]

for student in ai_students:
    student.showcase()
    print()
    student.full_report()
    print("-" * 50)
```

```python
# Exercise 9 — Vehicle inheritance
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def move(self):
        print(f"{self.brand} moves at {self.speed} km/h")

class Car(Vehicle):
    def __init__(self, brand, speed, fuel):
        super().__init__(brand, speed)
        self.fuel = fuel

    def honk(self):
        print(f"{self.brand} goes Beep Beep!")

class ElectricCar(Car):
    def __init__(self, brand, speed, battery_range):
        super().__init__(brand, speed, fuel="Electric")
        self.battery_range = battery_range

    def charge_status(self):
        print(f"{self.brand} | Range: {self.battery_range} km | Fuel: {self.fuel}")

c  = Car("Toyota", 120, "Petrol")
ec = ElectricCar("Tesla", 200, 500)

c.move()
c.honk()
ec.move()
ec.honk()
ec.charge_status()

# Practice  
class HybridCar(Car):
    def __init__(self, brand, speed, fuel, battery_range):
        super().__init__(brand, speed, fuel)
        self.battery_range = battery_range

    def charge_status(self):
        print(f"{self.brand} | Range: {self.battery_range} km | Fuel: {self.fuel} | Mode: Hybrid")

    def eco_drive(self):
        print(f"{self.brand} is running in eco mode for better mileage.")

hc = HybridCar("Honda", 150, "Hybrid", 650)

vehicles = [c, ec, hc]

for vehicle in vehicles:
    vehicle.move()
    if isinstance(vehicle, Car):
        vehicle.honk()
    if hasattr(vehicle, "charge_status"):
        vehicle.charge_status()
    if hasattr(vehicle, "eco_drive"):
        vehicle.eco_drive()
    print("-" * 40)
```

```python
# Exercise 11 — File not found handling
def read_file(filepath):
    try:
        with open(filepath, "r") as f:
            return f.read()
    except FileNotFoundError:
        print(f"Error: '{filepath}' not found")
        return None
    except PermissionError:
        print(f"Error: No permission to read '{filepath}'")
        return None
    finally:
        print(f"Attempted to read: {filepath}")

read_file("topics.txt")
read_file("missing_file.txt")

# Practice  
print("\nReading student scores file:")
scores_content = read_file("student_scores.txt")
if scores_content:
    print(scores_content)

print("\nReading prompt templates:")
templates_content = read_file("prompt_templates.txt")
if templates_content:
    print(templates_content)

print("\nAttempting to read non-existent model config:")
config_content = read_file("model_config.json")

print("\nReading session log:")
log_content = read_file("session_log.txt")
if log_content:
    lines = log_content.strip().split('\n')
    print(f"Total log entries: {len(lines)}")
    for entry in lines[:3]:
        print(f"  {entry}")
```

```python
# Exercise 12 — Simulated API call with error handling
import random

def call_model_api(model_name, tokens_needed):
    response_code = random.choice([200, 429, 500])
    if response_code == 429:
        raise ConnectionError(f"Rate limit exceeded for {model_name}")
    if response_code == 500:
        raise RuntimeError(f"Model server {model_name} is temporarily unavailable")
    return {"status": response_code, "model": model_name, "tokens_used": tokens_needed}

def safe_model_call(model_name, tokens_needed):
    try:
        result = call_model_api(model_name, tokens_needed)
        print(f"Success: {result['model']} processed {result['tokens_used']} tokens")
    except ConnectionError as e:
        print(f"Connection error: {e}")
    except RuntimeError as e:
        print(f"Runtime error: {e}")
    finally:
        print(f"Request attempt completed for model: '{model_name}'\n")

for model in ["llama3.2", "mistral", "qwen2.5:1.5b"]:
    tokens = token_map[model]
    safe_model_call(model, tokens)
```

```python
# Exercise 13 — Full PromptLogger class using everything from Day 1
from datetime import datetime

class PromptLogger:
    def __init__(self, model_name, log_file="prompt_log.txt"):
        self.model_name = model_name
        self.log_file = log_file
        self.history = []

    def send(self, prompt, temperature=0.7):
        entry = {
            "timestamp": datetime.now().strftime("%H:%M:%S"),
            "model": self.model_name,
            "prompt": prompt.strip(),
            "temperature": temperature,
            "tokens": len(prompt.split())
        }
        self.history.append(entry)
        print(f"[{entry['timestamp']}] Prompt sent: '{prompt}'")
        return entry

    def save_log(self):
        try:
            with open(self.log_file, "w") as f:
                for entry in self.history:
                    f.write(f"[{entry['timestamp']}] {entry['model']} | "
                            f"Temp: {entry['temperature']} | "
                            f"Tokens: {entry['tokens']} | "
                            f"{entry['prompt']}\n")
            print(f"\nLog saved to '{self.log_file}'")
        except Exception as e:
            print(f"Failed to save log: {e}")
        finally:
            print("Save attempt complete")

    def summary(self):
        print(f"\n--- Session Summary ({self.model_name}) ---")
        print(f"Total prompts : {len(self.history)}")
        print(f"Total tokens  : {sum(e['tokens'] for e in self.history)}")
        prompts = [e['prompt'] for e in self.history]
        print(f"Prompts sent  : {prompts}")

    def filtered(self, min_tokens=5):
        return [e for e in self.history if e["tokens"] >= min_tokens]


logger = PromptLogger("llama3.2")
logger.send("What is AI?")
logger.send("Explain backpropagation in simple terms", temperature=0.3)
logger.send("Give a real-world example of a transformer model")
logger.save_log()
logger.summary()

print(f"\nLong prompts: {[e['prompt'] for e in logger.filtered(min_tokens=6)]}")

# Practice 
print("\n" + "="*50)
logger_mistral = PromptLogger("mistral", log_file="mistral_log.txt")
logger_mistral.send("How does neural network training work?", temperature=0.5)
logger_mistral.send("Describe gradient descent", temperature=0.2)
logger_mistral.send("What are activation functions?")
logger_mistral.save_log()
logger_mistral.summary()

# Practice  
print("\n" + "="*50)
logger_qwen = PromptLogger("qwen2.5:1.5b", log_file="qwen_log.txt")
logger_qwen.send("Explain Django middleware", temperature=0.4)
logger_qwen.send("FastAPI vs Flask comparison")
logger_qwen.send("How to implement authentication in web frameworks?", temperature=0.6)
logger_qwen.save_log()
logger_qwen.summary()

print(f"\nQwen short prompts: {[e['prompt'] for e in logger_qwen.filtered(min_tokens=3)]}")
```
