```python
# Exercise 1 — Simplest class
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old.")

s1 = Student("Arun", 21)
s2 = Student("Priya", 20)

s1.introduce()
s2.introduce()
```

```python
# Exercise 2 — Class with multiple methods
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def result(self):
        return "Passed" if self.score >= 60 else "Failed"

    def grade(self):
        if self.score >= 90: return "S"
        elif self.score >= 80: return "A"
        elif self.score >= 70: return "B"
        elif self.score >= 60: return "C"
        else: return "F"

    def display(self):
        print(f"{self.name} | Score: {self.score} | Grade: {self.grade()} | {self.result()}")

s1 = Student("Arun", 88)
s2 = Student("Karan", 45)
s3 = Student("Priya", 95)

s1.display()
s2.display()
s3.display()
```

##Without OOP vs With OOP

Without OOP

```python
# All data stored as plain variables and lists
# All logic written as separate functions

student1_name = "Arun"
student1_score = 88
student1_subject = "Python"

student2_name = "Priya"
student2_score = 45
student2_subject = "Python"

def get_grade(score):
    if score >= 90:
        return "S"
    elif score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"

def get_result(score):
    return "Passed" if score >= 60 else "Failed"

def apply_curve(score, bonus):
    return score + bonus

def display_student(name, score, subject):
    grade  = get_grade(score)
    result = get_result(score)
    print(f"Name    : {name}")
    print(f"Subject : {subject}")
    print(f"Score   : {score}")
    print(f"Grade   : {grade}")
    print(f"Result  : {result}")
    print("---")

# Display
display_student(student1_name, student1_score, student1_subject)
display_student(student2_name, student2_score, student2_subject)

# Apply curve — have to manually pass every variable
student2_score = apply_curve(student2_score, 10)
print(f"After curve — {student2_name}: {student2_score}")
print(f"New Result  — {get_result(student2_score)}")
```

With OOP

```python
# Data and logic bundled together inside one class

class Student:
    def __init__(self, name, score, subject):
        self.name    = name
        self.score   = score
        self.subject = subject

    def get_grade(self):
        if self.score >= 90:
            return "S"
        elif self.score >= 80:
            return "A"
        elif self.score >= 70:
            return "B"
        elif self.score >= 60:
            return "C"
        else:
            return "F"

    def get_result(self):
        return "Passed" if self.score >= 60 else "Failed"

    def apply_curve(self, bonus):
        self.score += bonus
        print(f"Curve applied — New Score: {self.score}")

    def display(self):
        print(f"Name    : {self.name}")
        print(f"Subject : {self.subject}")
        print(f"Score   : {self.score}")
        print(f"Grade   : {self.get_grade()}")
        print(f"Result  : {self.get_result()}")
        print("---")


# Create objects — clean, readable, no variable mess
s1 = Student("Arun",  88, "Python")
s2 = Student("Priya", 45, "Python")

# Display
s1.display()
s2.display()

# Apply curve — object carries its own data, no manual passing
s2.apply_curve(10)
print(f"New Result — {s2.get_result()}")
```

Why python wins using OOPs

```python
# Adding more students is just one line each
s3 = Student("Karan", 72, "Python")
s4 = Student("Meena", 95, "Python")

# Store all students in a list and loop
students = [s1, s2, s3, s4]

for student in students:
    student.display()

# Filter — find all passed students
passed = [s for s in students if s.get_result() == "Passed"]
print(f"Passed students: {[s.name for s in passed]}")

# Apply curve to everyone who failed
for s in students:
    if s.get_result() == "Failed":
        s.apply_curve(10)
        print(f"{s.name} new result: {s.get_result()}")
```

## Real-World OOP Examples

```python
# Exercise 3 — BankAccount class
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.transactions = []

    def deposit(self, amount):
        self.balance += amount
        self.transactions.append(f"Deposit  : +{amount}")
        print(f"Deposited ₹{amount}. Balance: ₹{self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance -= amount
            self.transactions.append(f"Withdraw : -{amount}")
            print(f"Withdrew ₹{amount}. Balance: ₹{self.balance}")

    def statement(self):
        print(f"\n--- Statement for {self.owner} ---")
        for t in self.transactions:
            print(f"  {t}")
        print(f"  Final Balance: ₹{self.balance}")

acc = BankAccount("Arun", 1000)
acc.deposit(500)
acc.withdraw(200)
acc.withdraw(2000)
acc.statement()
```

```python
# Exercise 4 — Library book management
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

    def borrow(self, student_name):
        if self.is_available:
            self.is_available = False
            print(f"'{self.title}' borrowed by {student_name}")
        else:
            print(f"'{self.title}' is not available")

    def return_book(self):
        self.is_available = True
        print(f"'{self.title}' returned. Available again.")

    def status(self):
        state = "Available" if self.is_available else "Borrowed"
        print(f"'{self.title}' by {self.author} — {state}")

b1 = Book("Deep Learning", "Goodfellow")
b2 = Book("Python Crash Course", "Matthes")

b1.status()
b1.borrow("Priya")
b1.borrow("Karan")
b1.return_book()
b1.borrow("Karan")
b2.status()
```

```python
# Exercise 5 — Simple AI Chatbot session class
class ChatSession:
    def __init__(self, model_name):
        self.model_name = model_name
        self.messages = []
        self.token_count = 0

    def send(self, prompt):
        tokens = len(prompt.split())
        self.token_count += tokens
        self.messages.append({"role": "user", "content": prompt})
        print(f"[{self.model_name}] Prompt received: '{prompt}'")

    def get_history(self):
        print(f"\n--- Chat History ({self.model_name}) ---")
        for i, msg in enumerate(self.messages, start=1):
            print(f"  {i}. {msg['role'].upper()}: {msg['content']}")

    def stats(self):
        print(f"\nTotal messages : {len(self.messages)}")
        print(f"Total tokens   : {self.token_count}")

session = ChatSession("llama3.2")
session.send("What is AI?")
session.send("Explain machine learning")
session.send("What is backpropagation?")
session.get_history()
session.stats()
```

```python
# Exercise 6 — Todo task manager
class TaskManager:
    def __init__(self, owner):
        self.owner = owner
        self.tasks = []

    def add(self, task):
        self.tasks.append({"task": task, "done": False})
        print(f"Added: '{task}'")

    def complete(self, index):
        self.tasks[index]["done"] = True
        print(f"Completed: '{self.tasks[index]['task']}'")

    def pending(self):
        return [t for t in self.tasks if not t["done"]]

    def display(self):
        print(f"\n--- Tasks for {self.owner} ---")
        for i, t in enumerate(self.tasks):
            status = "✓" if t["done"] else "○"
            print(f"  [{status}] {i}. {t['task']}")

tm = TaskManager("Arun")
tm.add("Complete Python exercises")
tm.add("Watch ML lecture")
tm.add("Set up VS Code")
tm.complete(0)
tm.complete(2)
tm.display()
print(f"\nPending: {[t['task'] for t in tm.pending()]}")
```

## Inheritance Examples

```python
# Exercise 7 — Base class and child classes
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def speak(self):
        print(f"{self.name} says {self.sound}")

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name, "Woof")

    def fetch(self):
        print(f"{self.name} fetches the ball!")

class Cat(Animal):
    def __init__(self, name):
        super().__init__(name, "Meow")

    def purr(self):
        print(f"{self.name} purrs...")

d = Dog("Bruno")
c = Cat("Whiskers")

d.speak()
d.fetch()
c.speak()
c.purr()
```

```python
# Exercise 8 — Student → AIStudent inheritance
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def result(self):
        return f"{self.name}: {'Passed' if self.score >= 60 else 'Failed'}"

class AIStudent(Student):
    def __init__(self, name, score, project, model_used):
        super().__init__(name, score)
        self.project = project
        self.model_used = model_used

    def showcase(self):
        print(f"{self.name} built '{self.project}' using {self.model_used}")

    def full_report(self):
        print(f"Name      : {self.name}")
        print(f"Score     : {self.score}")
        print(f"Result    : {self.result()}")
        print(f"Project   : {self.project}")
        print(f"Model     : {self.model_used}")

s = AIStudent("Priya", 95, "RAG Chatbot", "llama3.2")
s1 = AIStudent("Ravi", 90, "RAG Chatbot", "Gemma")
s.showcase()
s.full_report()

s1.showcase()
s1.full_report()
```

```python
# Exercise 9 — Vehicle inheritance
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def move(self):
        print(f"{self.brand} moves at {self.speed} km/h")

class Car(Vehicle):
    def __init__(self, brand, speed, fuel):
        super().__init__(brand, speed)
        self.fuel = fuel

    def honk(self):
        print(f"{self.brand} goes Beep Beep!")

class ElectricCar(Car):
    def __init__(self, brand, speed, battery_range):
        super().__init__(brand, speed, fuel="Electric")
        self.battery_range = battery_range

    def charge_status(self):
        print(f"{self.brand} | Range: {self.battery_range} km | Fuel: {self.fuel}")

c  = Car("Toyota", 120, "Petrol")
ec = ElectricCar("Tesla", 200, 500)

c.move()
c.honk()
ec.move()
ec.honk()
ec.charge_status()
```

## try / except / finally

```python
# Exercise 10 — Handle division errors
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

```python
# Exercise 11 — File not found handling
def read_file(filepath):
    try:
        with open(filepath, "r") as f:
            return f.read()
    except FileNotFoundError:
        print(f"Error: '{filepath}' not found")
        return None
    except PermissionError:
        print(f"Error: No permission to read '{filepath}'")
        return None
    finally:
        print(f"Attempted to read: {filepath}")

read_file("topics.txt")
read_file("missing_file.txt")
```

```python
# Exercise 12 — Simulated API call with error handling
import random

def call_api(prompt):
    status = random.choice([200, 429, 500])
    if status == 429:
        raise ConnectionError("Rate limit exceeded")
    if status == 500:
        raise RuntimeError("Server error")
    return {"status": status, "response": f"Answer to: {prompt}"}

def safe_api_call(prompt):
    try:
        result = call_api(prompt)
        print(f"Success: {result['response']}")
    except ConnectionError as e:
        print(f"Connection error: {e}")
    except RuntimeError as e:
        print(f"Runtime error: {e}")
    finally:
        print(f"Attempt complete for: '{prompt}'\n")

for prompt in ["What is AI?", "Explain ML", "Define NLP"]:
    safe_api_call(prompt)
```

## Putting It All Together

```python
# Exercise 13 — Full PromptLogger class using everything from Day 1
from datetime import datetime

class PromptLogger:
    def __init__(self, model_name, log_file="prompt_log.txt"):
        self.model_name = model_name
        self.log_file = log_file
        self.history = []

    def send(self, prompt, temperature=0.7):
        entry = {
            "timestamp": datetime.now().strftime("%H:%M:%S"),
            "model": self.model_name,
            "prompt": prompt.strip(),
            "temperature": temperature,
            "tokens": len(prompt.split())
        }
        self.history.append(entry)
        print(f"[{entry['timestamp']}] Prompt sent: '{prompt}'")
        return entry

    def save_log(self):
        try:
            with open(self.log_file, "w") as f:
                for entry in self.history:
                    f.write(f"[{entry['timestamp']}] {entry['model']} | "
                            f"Temp: {entry['temperature']} | "
                            f"Tokens: {entry['tokens']} | "
                            f"{entry['prompt']}\n")
            print(f"\nLog saved to '{self.log_file}'")
        except Exception as e:
            print(f"Failed to save log: {e}")
        finally:
            print("Save attempt complete")

    def summary(self):
        print(f"\n--- Session Summary ({self.model_name}) ---")
        print(f"Total prompts : {len(self.history)}")
        print(f"Total tokens  : {sum(e['tokens'] for e in self.history)}")
        prompts = [e['prompt'] for e in self.history]
        print(f"Prompts sent  : {prompts}")

    def filtered(self, min_tokens=5):
        return [e for e in self.history if e["tokens"] >= min_tokens]


logger = PromptLogger("llama3.2")
logger.send("What is AI?")
logger.send("Explain backpropagation in simple terms", temperature=0.3)
logger.send("Give a real-world example of a transformer model")
logger.save_log()
logger.summary()

print(f"\nLong prompts: {[e['prompt'] for e in logger.filtered(min_tokens=6)]}")
```
